```python
#Scatter plot of age and employee salary
plt.figure(figsize=(8,6))
sns.scatterplot(data=data,x='age',y='salary',color='red',edgecolor='linen',alpha=0.5)
plt.title("Scatter plot of age and employee salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()

```

```python
#Scatterplot of healthy_eating and active_lifestyle
plt.figure(figsize=(8,6))
sns.scatterplot(data=data,x='healthy_eating',y='active_lifestyle',color='blue',edgecolor='linen',alpha=0.5)
plt.title("Scatter plot of healthy_eating and active_lifestyle")
plt.xlabel("healthy_eating")
plt.ylabel("active_lifestyle")
plt.show()
```

```python
#Scatterplot for healthy_eating and salary
plt.figure(figsize=(8,6))
sns.scatterplot(data=data,x='healthy_eating',y='salary',color='green',edgecolor='linen',alpha=0.5)
plt.title("Scatter plot of healthy_eating and salary")
plt.xlabel("healthy_eating")
plt.ylabel("salary")
plt.show()
```

```python
#Create a count plot of the groups
plt.figure(figsize=(8,6))
sns.countplot(data=data,x='groups',edgecolor='black',alpha=0.7)
plt.title("Count plot of groups")
plt.xlabel("Groups")
plt.ylabel("Count")
plt.show()
```

```python
#Histogram of salary
plt.figure(figsize=(8,6))
sns.histplot(data=data,x='salary',color='yellow',edgecolor='black',alpha=0.7,bins=10)
plt.title("Histogram of salary")
plt.xlabel("Salary")
plt.ylabel("Count")
```
